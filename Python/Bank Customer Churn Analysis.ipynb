## Step 8 — Statistical Analysis

```python
## IQR → Outlier detection

numeric_cols=['credit_score','age','tenure','balance','products_number','estimated_salary']

Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outlier_count = ((df[numeric_cols] < lower_limit) |
                 (df[numeric_cols] > upper_limit)).sum()

outlier_count
```

Conclusion:
IQR analysis identified some extreme observations, particularly in age and products number. These observations were investigated and retained because they can represent valid customer characteristics.

```python
## T-test → Compare 2 groups

from scipy.stats import ttest_ind

stayed_age = df[df['churn'] == 0]['age']
churned_age = df[df['churn'] == 1]['age']

t_stat, p_value = ttest_ind(stayed_age, churned_age)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

We use:

H₀: Mean age is the same for both groups.

H₁: Mean age is different between the two groups.

If:

p-value < 0.05

→ Reject H₀ → statistically significant difference.

If:

p-value >= 0.05

→ Fail to reject H₀ → no statistically significant difference.

Conclusion

We'll write the exact conclusion after you get your p-value.

```python
## ANOVA → Compare 3+ groups

from scipy.stats import f_oneway

france_age = df[df['country'] == 'France']['age']
germany_age = df[df['country'] == 'Germany']['age']
spain_age = df[df['country'] == 'Spain']['age']

f_stat, p_value = f_oneway(
    france_age,
    germany_age,
    spain_age
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

Hypotheses

H₀: Mean age is the same across all countries.

H₁: At least one country's mean age is different.

Again:

p < 0.05 → significant

p ≥ 0.05 → not significant

```python
## Chi-square → Relationship between categorical variables

from scipy.stats import chi2_contingency

table = pd.crosstab(df['country'], df['churn'])
table
```

```python
chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-square:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
```

Hypotheses

H₀: Country and churn are independent.

H₁: Country and churn are associated.

Interpretation

If:

p-value < 0.05

→ There is a statistically significant association between country and churn.

If:

p-value ≥ 0.05

→ We don't have enough evidence to say they are associated.
